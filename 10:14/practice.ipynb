##### No.3757 Happy End

```python

N = int(input()) 
S = input()
answer = "HAPPY" 

last_five = S[-5:] # 最後の5文字
other = S[:-5] # 最後の５文字以外
times = 0 

# 末尾以外のHAPPYの出現回数をカウントする
while answer in other:
    times += 1
    other = other.replace(answer, "*", 1)

for i in range(5):
    if last_five[i] != answer[i]:
        times += 1 

print(times)
```

##### No.3721 Absurd Basic Constructive

```python
import random

#ランダムな　N＊N のボードを作成する
N , K = map (int, input().split())
numbers = list(range(1, N**2 + 1))
random.shuffle(numbers)
board = [numbers[i*N:(i+1)*N] for i in range(N)]

#N＊N のボードからすべての長方形を列挙する
def get_rectangles(board):
    N = len(board)
    rectangles = []

    for i1 in range(N):
        for j1 in range(N):
            for i2 in range(i1, N):
                for j2 in range(j1, N):
                    rectangle = [
                    [board[i][j] for i in range (i1, i2+1)]
                    for j in range (j1, j2+1)
                ]
                    rectangles.append(rectangle)
    return rectangles

#各長方形の最大値を確認し、条件を満たす数を数える
def count_rectangles(board):
    count = 0
    rectangles = get_rectangles (board)

    for rectangle in rectangles:
        max_value = max(max(row) for row in rectangle) 
        
    if max_value == board[-1][-1]: 
        count += 1

    return count

count = count_rectangles(board)

if count < K:
    print (-1)






if count == K:
    print (board)
```
