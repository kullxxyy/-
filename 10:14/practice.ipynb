##### No.3757 Happy End

```python

N = int(input()) 
S = input()
answer = "HAPPY" 

last_five = S[-5:] # 最後の5文字
other = S[:-5] # 最後の５文字以外
times = 0 

# 末尾以外のHAPPYの出現回数をカウントする
while answer in other:
    times += 1
    other = other.replace(answer, "*", 1)

for i in range(5):
    if last_five[i] != answer[i]:
        times += 1 

print(times)
```

##### No.3721 Absurd Basic Constructive

```python
import random

N , K = map (int, input().split())
count = 0
numbers = list(range(1, N**2 + 1))
random.shuffle(numbers)
board = [numbers[i*N:(i+1)*N] for i in range(N)]

for i in range (1, N+1):
    for j in range (1, N+1):
        if board[i][j] > max(board[i-1][j], board[i][j-1]):
            count += 1

if count == K:
    print (board)
```
