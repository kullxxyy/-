##### No.3757 Happy End

```python

N = int(input()) 
S = input()
answer = "HAPPY" 

last_five = S[-5:] # 最後の5文字
other = S[:-5] # 最後の５文字以外
times = 0 

# 末尾以外のHAPPYの出現回数をカウントする
while answer in other:
    times += 1
    other = other.replace(answer, "*", 1)

for i in range(5):
    if last_five[i] != answer[i]:
        times += 1 

print(times)
```
