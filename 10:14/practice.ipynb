##### No.3757 Happy End

```python

N = int(input()) 
S = input()
answer = "HAPPY" 

last_five = S[-5:] # 最後の5文字
other = S[:-5] # 最後の５文字以外
times = 0 

# 末尾以外のHAPPYの出現回数をカウントする
while answer in other:
    times += 1
    other = other.replace(answer, "*", 1)

for i in range(5):
    if last_five[i] != answer[i]:
        times += 1 

print(times)
```

##### No.3721 Absurd Basic Constructive

```python
from itertools import permutations

# 数字を重複せずに入れる
def fill_board(board, N, numbers):
    
    for row in range(N):
        for col in range(N):
            board[row][col] = numbers[row*N + col]

    return board


#N＊N のボードからすべての長方形を列挙する
def get_rectangles(board, N):
    rectangles = []
    start_row = 0
    start_col = 0
    for end_row in range(start_row, N):
        for end_col in range(start_col, N):
            rectangle = [
                [board[row][col] for col in range(start_col, end_col+1)]
                    for row in range(start_row, end_row+1)
                ]
            rectangles.append(rectangle)
    return rectangles

#各長方形の最大値を確認し、条件を満たす数を数える
def count_rectangles(board):
    count = 0
    rectangles = get_rectangles (board, len(board))
    for rectangle in rectangles:
        max_value = max(max(row) for row in rectangle) 
        if max_value == rectangle[-1][-1]: 
            count += 1
    return count


def main() :
    N = 3
    K = int(input())

    numbers = list(range(1, N**2 + 1))
    board = [
    [0 for col in range(N)]
    for row in range(N)
    ]

    # すべての数字の配置を調べ、条件を満たすボードを探す
    for p in permutations(numbers):
        #　数字を複数せずに入れる
        board = fill_board(board, N, p)

        # 各長方形の最大値を確認し、条件を満たす数を数える
        count = count_rectangles(board)

        #　終了条件のチェック
        if count == K:
            for row in board:
                print(*row)
            break
    else:
        print(-1)

if __name__ == "__main__":
    main()




```
