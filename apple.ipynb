```python
#変数定義
answer ="apple"
misses = 0
max_misses = 6
guessed_letters = []
```

```python
def check_guess(guess):
    if len(guess) == 1 and guess.isalpha() and guess not in guessed_letters:
        if guess in answer:
            print("correct")
        else:
            print("incorrect")
            misses += 1
        guessed_letters.append(guess)
    else:
        print("Invalid input. Please try again.")
```

```python
#ループ処理
for i in answer: 
    if i in guessed_letters:
        print(i, end="")
    else:
        print("_", end="")

```
