#### ハングマンゲーム

##### 1. 変数の定義

```python
answer ="apple"
misses = 0
max_misses = 6
guessed_letters = []
```

##### 2. 初期画面とプレイヤーの入力

```python
def show_word(answer, guessed_letters):
    for i in answer: 
        if i in guessed_letters:
            print(i, end=" ")
        else:
            print("_", end=" ")
print("Word:", end=" ")
show_word(answer, guessed_letters)
print()
print("Misses:", misses, "/", max_misses)
input("Please enter a letter: ")
```

##### 3. 入力内容の実現可能性の判断

```python
def get_valid_guess():
    while True:
        letter = input("Please enter a letter: ")

        if len(letter) == 1 and letter.isalpha() and letter not in guessed_letters and letter.isascii():
            return letter
        else:
            print("Invalid input. Please try again.")

guess = get_valid_guess()
```

##### 4. 入力内容の正確性の判定

```python
def check_guess(guess, answer, misses, guessed_letters):
    if guess in answer:
        print("correct")
    else:
        print("incorrect")
        misses += 1
    guessed_letters.append(guess)
```
