#### ハングマンゲーム

##### 1. 変数の定義

```python
answer ="apple"
misses = 0
max_misses = 6
guessed_letters = []
```

##### 2. 初期画面とプレイヤーの入力

```python
def show_word(answer, guessed_letters):
    for i in answer: 
        if i in guessed_letters:
            print(i, end=" ")
        else:
            print("_", end=" ")
print("Word:", end=" ")
show_word(answer, guessed_letters)
print()
print("Misses:", misses, "/", max_misses)
```

##### 3. 入力内容の実現可能性の判断

```python
def get_valid_guess():
    while True:
        letter = input("Please enter a letter: ")

        if len(letter) == 1 and letter.isalpha() and letter not in guessed_letters and letter.isascii():
            return letter
        else:
            print("Invalid input. Please try again.")
```

##### 4. 入力内容の正確性の判定

```python
def check_guess(guess, answer, misses):
    if guess in answer:
        pass
    else:
        misses += 1
    return misses
```

##### 5. 勝利の判断

#### メインプログラム / ゲームループ：

```python
# まずはゲーム開始時の状態を表示します。
print("Word:", end=" ")
show_word(answer, guessed_letters)
print()
print("Misses:", misses, "/", max_misses)

while True:
    # 1 プレイヤーの入力
    guess = get_valid_guess()

    # 2 推測した文字を記録する
    guessed_letters.append(guess)

    # 3 正誤を判定し、missesを更新する
    misses = check_guess(guess, answer, misses)

    # 4 このラウンド終了後の結果を表示する
    print("Word:", end=" ")
    show_word(answer, guessed_letters)
    print()
    print("Misses:", misses, "/", max_misses)

    # 5 ゲーム終了条件のチェック
    if misses >= max_misses:
        print("You lose!")
        break

    if all(letter in guessed_letters for letter in answer):
        print("You win!")
        break
```
